# Summary of the KNN algrorithms benchmarks

To build k-nearest neighbor (KNN) graphs, we benchmark two methods: a custom pure PyTorch implementation and the FAISS library.

The benchmarks are performed on several architectures. For GPUs, here is the list of tested devices:

* [NVIDIA RTX 4000 Ada Generation (12GB)](#nvidia-rtx-4000-ada-generation) on a local laptop
* [NVIDIA V100 (16GB)](#nvidia-v100-16gb) at the [Jean Zay supercomputer](https://www.idris.fr/jean-zay/)
* [NVIDIA V100 (32GB)](#nvidia-v100-32gb) at the [Jean Zay supercomputer](https://www.idris.fr/jean-zay/) and at the [IN2P3 Computing Center](https://cc.in2p3.fr/)
* [NVIDIA H100 (80GB)](#nvidia-h100-80gb) at the [Jean Zay supercomputer](https://www.idris.fr/jean-zay/)
* [NVIDIA H100 (94GB)](#nvidia-h100-94gb) at the [IN2P3 Computing Center](https://cc.in2p3.fr/)

For CPUs, tests have only been performed locally on an Intel i9-13950HX.

## Results

### GPU benchmarks

#### NVIDIA RTX 4000 Ada Generation

![KNN Benchmark on NVIDIA RTX 4000 Ada Generation](../tests/knn_benchmarks/laptop_DB-torch-faiss-memory_and_runtime_vs_system_size.png)

We see that for small systems (up to a few thousands of atoms), the PyTorch implementation is much faster than FAISS. However, when we reach 2.5k atoms, the cost of PyTorch increases steeply, until becoming slower than FAISS for systems larger than ~5k atoms.

On the memory side, FAISS is much more efficient, with a memory usage which is a lot smaller than PyTorch. The increase in memory needed with system size is also much more gradual for FAISS. While PyTorch quickly reaches the 12GB limit of the GPU (at ~6k atoms), FAISS barely needs 0.01GB for all tested system sizes (up to 20k atoms). 

#### NVIDIA V100 (16GB)

![KNN Benchmark on NVIDIA V100 (16GB)](../tests/knn_benchmarks/JZ-V100-16G-torch-faiss-memory_and_runtime_vs_system_size.png)

The behaviour is qualitatively similar to the one observed on the RTX 4000 Ada Generation. The values for crossover for runtimes shift a bit:

* the steep increase in PyTorch runtime happens at similar system sizes (~2.5k atoms)
* the crossover point where FAISS becomes faster than PyTorch happens at a slightly larger system size (~6k atoms instead of ~5k atoms)
* the maximum system size that can be tested with PyTorch is larger (~7k atoms instead of ~6k atoms), thanks to the larger memory available on the GPU (16GB instead of 12GB)

#### NVIDIA V100 (32GB)

At Jean Zay:

![KNN Benchmark on NVIDIA V100 (32GB) at Jean-Zay](../tests/knn_benchmarks/JZ-V100-32G-torch-faiss-memory_and_runtime_vs_system_size.png)

At IN2P3 Computing Center:

![KNN Benchmark on NVIDIA V100 (32GB) at IN2P3](../tests/knn_benchmarks/IN2P3-V100-torch-faiss-memory_and_runtime_vs_system_size.png)

For those two different facilities, using the same GPU model yields very similar results (which is of course expected, but still reassuring). The behaviour is again qualitatively similar to the one observed on the RTX 4000 Ada Generation and on the V100 (16GB). The values for crossover for runtimes shift a bit again:

* the steep increase in PyTorch runtime happens a bit later (~5k atoms instead of ~2.5k atoms)
* the crossover point where FAISS becomes faster than PyTorch happens at a larger system size (~7.5k atoms instead of ~6k atoms)
* the maximum system size that can be tested with PyTorch thanks to the memory doubling is larger (~10k atoms instead of ~7k atoms).

#### NVIDIA H100 (80GB)

![KNN Benchmark on NVIDIA H100 (80GB)](../tests/knn_benchmarks/JZ-H100-torch-faiss-memory_and_runtime_vs_system_size.png)

The behaviour is again qualitatively similar to the one observed on the RTX 4000 Ada Generation and on the V100 (16GB and 32GB). The values for crossover for runtimes shift a bit again:

* the steep increase in PyTorch runtime happens again around ~5k atoms
* the crossover point where FAISS becomes faster than PyTorch happens at a slightly larger system size (~10k atoms instead of ~7.5k atoms)
* the maximum system size that can be tested with PyTorch thanks to the larger memory is larger (~16k atoms instead of ~10k atoms).

#### NVIDIA H100 (94GB)

![KNN Benchmark on NVIDIA H100 (94GB)](../tests/knn_benchmarks/IN2P3-H100-torch-faiss-memory_and_runtime_vs_system_size.png)

The behaviour is again qualitatively similar to the one observed on the RTX 4000 Ada Generation and on the V100 (16GB and 32GB) and on the H100 (80GB). The values for crossover for runtimes shift a bit again:

* the steep increase in PyTorch runtime happens now around 10k atoms, although a slower increase seems to start around ~1k atoms
* the crossover point where FAISS becomes faster than PyTorch happens around the same system size as on the H100 (80GB) (~10k atoms)
* the maximum system size that can be tested with PyTorch thanks to the larger memory is larger (~17.5k atoms instead of ~16k atoms).

## CPU benchmarks

### Intel i9-13950HX

![KNN Benchmark on Intel i9-13950HX](../tests/knn_benchmarks/laptop_DB-CPU-torch-faiss-runtime_vs_system_size.png)

> **Note:** Here, no memory benchmark is shown since it is not as straightforward to measure memory usage on CPU as on GPU. 

The overall cost in this case is different, with the runtime scaling almost exponentially (mind the logscale on the y-axis) with system size for both tested implementations. The PyTorch implementation is always a little faster than FAISS, but the evolution with system size is roughly the same for both methods. Therefore, there is no crossover point as observed for GPUs.

Also, the total runtime is smaller for small systems (up to 10 atoms) with both methods compared to the GPU benchmarks, but it quickly becomes much more prohibitive for larger systems. For instance, systems with 5k atoms require ~1s to ~10s on CPU, and only ~0.1s on GPU. For 10k atoms (when fitting in the GPU memory is possible), the CPU runtime is ~10s to ~20s while it is still a lot below 1s on GPU.

For this reason, I have not attempted to benchmark systems larger than 12.5k atoms on CPU.
