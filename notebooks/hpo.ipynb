```python
def objective_vs_time(trial: optuna.trial.FrozenTrial):
    objective = trial.value
    duration_s = trial.duration.total_seconds()

    assert objective is not None
    assert duration_s is not None

    return objective, duration_s / 60


optuna.visualization.plot_pareto_front(
    study, targets=objective_vs_time, target_names=["Objective Value", "Training Time (min)"]
).show()
```
