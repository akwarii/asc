```python
import optuna
```

```python
STUDY_NAMES = ["mlp", "cegann"]
STORAGE = f"sqlite:///../logs/{STUDY_NAMES}.db"
NEW_STORAGE = "sqlite:///../logs/hpo.db"
```

```python
for study_name in STUDY_NAMES:
    old_storage = f"sqlite:///../logs/{study_name}.db"

    print(f"Copying trials from {old_storage} to {NEW_STORAGE} for study {study_name}")
    old_study = optuna.load_study(study_name=study_name, storage=old_storage)
    new_study = optuna.create_study(
        study_name=study_name,
        storage=NEW_STORAGE,
        load_if_exists=True,
        direction="maximize",
    )

    new_study.add_trials(old_study.get_trials())
```
