```python
from itertools import product

import optuna
```

```python
STORAGE_NAMES = ["hpo-in2p3", "hpo-idris"]
STUDY_NAMES = ["mlp", "cegann", "gat"]
NEW_STORAGE = "sqlite:///../logs/hpo.db"
```

```python
for current_storage, study in product(STORAGE_NAMES, STUDY_NAMES):
    old_study = optuna.load_study(
        study_name=study,
        storage=f"sqlite:///../logs/{current_storage}.db",
    )

    new_study = optuna.create_study(
        study_name=study,
        storage=NEW_STORAGE,
        load_if_exists=True,
        direction="maximize",
    )

    new_study.add_trials(old_study.get_trials())
```
